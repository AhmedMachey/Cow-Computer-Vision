```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\\Users\\AHMED\\OneDrive\\Desktop\\projet\\src\\vache1.jpg")

if img is None:
    print("Image introuvable")
    exit()

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_blur = cv2.GaussianBlur(gray, (5, 5), 0)

clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
gray_eq = clahe.apply(gray_blur)
# Affichage pour vérifier
plt.figure(figsize=(12, 4))
plt.subplot(1,3,1); plt.imshow(gray, cmap='gray'); plt.title("Grayscale")
plt.subplot(1,3,2); plt.imshow(gray_blur, cmap='gray'); plt.title("Filtré")
plt.subplot(1,3,3); plt.imshow(gray_eq, cmap='gray'); plt.title("CLAHE")
plt.show()
```

```python
h, w = img.shape[:2]

# Masques pour GrabCut
mask = np.zeros((h, w), np.uint8)
bgdModel = np.zeros((1, 65), np.float64)
fgdModel = np.zeros((1, 65), np.float64)

# Rectangle initial
rect = (10, 10, w - 20, h - 20)

# GrabCut (5 itérations)
cv2.grabCut(img, mask, rect, bgdModel, fgdModel, 5, cv2.GC_INIT_WITH_RECT)

# Conversion en masque binaire
mask_bin = np.where((mask == 0) | (mask == 2), 0, 255).astype("uint8")

plt.figure(figsize=(10,5))
plt.title("Masque GrabCut (binaire)")
plt.imshow(mask_bin, cmap='gray')
plt.axis("off")
plt.show()
```

```python
# Morphologie
kernel = np.ones((15, 15), np.uint8)
mask_clean = cv2.morphologyEx(mask_bin, cv2.MORPH_CLOSE, kernel, iterations=2)
mask_clean = cv2.morphologyEx(mask_clean, cv2.MORPH_OPEN, kernel, iterations=2)

# Recherche des contours
contours, _ = cv2.findContours(mask_clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contours = [c for c in contours if cv2.contourArea(c) > 1000]

largest = max(contours, key=cv2.contourArea)

# Affichage du masque nettoyé
plt.figure(figsize=(10,5))
plt.title("Masque après morphologie")
plt.imshow(mask_clean, cmap='gray')
plt.axis("off")
plt.show()
```

```python
surface_px = cv2.contourArea(largest)

# Rectangle englobant
x, y, w2, h2 = cv2.boundingRect(largest)
image_contours = img.copy()

cv2.drawContours(image_contours, [largest], -1, (0, 0, 255), 3)
cv2.rectangle(image_contours, (x, y), (x + w2, y + h2), (255, 0, 0), 2)

plt.figure(figsize=(10,5))
plt.title("Silhouette extraite (Contour + Bounding Box)")
plt.imshow(cv2.cvtColor(image_contours, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
COEFF_A = 0.0025
COEFF_B = 50.0
POIDS_SEUIL = 650
poids = COEFF_A * surface_px + COEFF_B
classification = "Vache a viande" if poids > POIDS_SEUIL else "Vache laitiere"

image_final = img.copy()

cv2.drawContours(image_final, [largest], -1, (0, 0, 255), 3)
cv2.rectangle(image_final, (x, y), (x + w2, y + h2), (255, 0, 0), 2)

cv2.putText(image_final, f"Poids : {poids:.1f} kg", (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

cv2.putText(image_final, f"Type : {classification}", (30, 100),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

plt.figure(figsize=(12,6))
plt.title(f"{classification} - Poids estimé : {poids:.1f} kg")
plt.imshow(cv2.cvtColor(image_final, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()

```
